```python
import numpy as np
import pandas as pd

df = pd.read_csv("housing.csv")

# Create income categories
df["income_cat"] = pd.cut(
    df["median_income"],
    bins=[0, 1.5, 3.0, 4.5, 6.0, np.inf],
    labels=[1, 2, 3, 4, 5]
)
```

```python
from sklearn.model_selection import StratifiedShuffleSplit

# Assume income_cat is a column in the dataset created from median_income
split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)

for train_index, test_index in split.split(df, df["income_cat"]):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]
```

```python
strat_train_set
```

```python
#Lets remove the income_cat column
for sett in (strat_train_set, strat_test_set):
    sett.drop("income_cat", axis = 1, inplace = True)
    
```

```python
df = strat_train_set.copy()
```

```python
housing = strat_train_set.drop("median_house_value", axis = 1)
housing_labels = strat_train_set["median_house_value"].copy()
```

```python
#we will handle the missing data
# 1. Drop rows with missing values. 2. Drop the Entire Column 3. Impute missing values(recommended)
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy = "median") #Now the imputer is ready to fit in the housing num
```

```python
#I want only numerical features in my data but there is ocean proximity habing strings 
housing_num = housing.select_dtypes(include=[np.number])
```

```python
imputer.fit(housing_num) #imputer is ready to impute it will replace NA by most probable value
```

```python
#it stores the medians in imputer.statistics_
```

```python
X = imputer.transform(housing_num) # x is the complete data with imputed medians
```
